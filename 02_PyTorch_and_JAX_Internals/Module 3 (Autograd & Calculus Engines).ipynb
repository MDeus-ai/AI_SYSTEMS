```python
import torch
```

```python
x = torch.tensor([2.0], requires_grad=True)
w = torch.tensor([3.0], requires_grad=True)
b = torch.tensor([1.0], requires_grad=True)
```

```python
# The Forward Pass
y = x * w + b

print(f"y: {y}")
print(f"y's Gradient Function: {y.grad_fn}")
print(f"Is y a leaf node? {y.is_leaf}")
print(f"Is x a leaf node? {x.is_leaf}")
```
