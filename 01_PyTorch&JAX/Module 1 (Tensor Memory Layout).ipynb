```python
# Indexing into storage
# A storage is a one-dimensional array of numerical data: that is, a contiguous block of memory containing numbers of a given type, such as `float` or `int64`
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points.storage()
```

```python
# Index into a storage manually
points_storage = points.storage()
points_storage[0]
```

```python
# Changing the value of a storage leads to changing the content of its referring tensor
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points_storage = points.storage()
points_storage[0] = 2.1

print(points)
```
